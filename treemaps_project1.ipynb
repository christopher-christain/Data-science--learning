```python
import pandas as pd
import plotly.express as px
```

```python
data = {
    'categories': ['Electronics', 'Electronics', 'Electronics',
    'Furniture', 'Furniture', 'Furniture',
                   'Clothing', 'Clothing', 'Clothing'],
    'subcategories': ['laptops', 'smartphones', 'tablets',
                      'chairs', 'tables', 'sofas',
                      'men', 'women', 'kids'],
    'values': [120000, 80000, 30000,
               50000, 40000, 20000,
               70000, 90000, 40000]
}
```

```python

df = pd.DataFrame(data)
```

```python
fig = px.treemap(df, path=['categories', 'subcategories'],
                 values = 'values',
                 title= 'sales treemap')
fig.show()
```
