```python
import pandas as pd
import plotly.express as px
```

```python
data = {
    'Categories': ['Category 1', 'Category 2', 'Category 3', 'Category 3', 'Category 4'],
    'Subcategories': ['Subcategory 1A', 'Subcategory 1B', 'Subcategory 2A', 'Subcategory 2B', 'Subcategory 3A'],
    'Values' : [10, 20, 30 , 40, 50]
}
```

```python
df = pd.DataFrame(data)
```

```python
display(df)
```

```python
fig = px.treemap(df,
                 path = ['Categories', 'Subcategories'],
                 values = 'Values',
                 title= 'Treemap Examples')
```

```python
fig.show()
```
